```
from PIL import Image
```

```
import torchvision.transforms as T
```

```
import matplotlib.pyplot as plt
```

```
# Define your candidate transforms
```

```
aug_transforms = T.Compose([ T.Resize((256, 256)), T.RandomResizedCrop(224, scale=(0.8, 1.0)), T.RandomHorizontalFlip(), T.ColorJitter(brightness=0.2, contrast=0.2), T.RandomRotation(15), ])
```

```
# Load a sample image
```

```
img = Image.open('data/images/YOUR_SAMPLE_IMAGE.jpg')
```

```
# Plot augmented versions
```

```
fig, axes = plt.subplots(4, 4, figsize=(12, 12))
```

```
for ax in axes.flatten(): ax.imshow(aug_transforms(img)); ax.axis('off')
```
